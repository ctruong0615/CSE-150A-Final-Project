# Steam data preview

Open this notebook from the project folder in Jupyter or VS Code and select a Python 3.10+ kernel. Run the cells in order, or choose **Run All**.

The first run downloads Kaggle dataset version 59 (about 401 MB extracted) and prepares the data. Later runs reuse the local files. An internet connection is needed for the download.

The preview uses **pandas**. If it is missing from your notebook kernel, run `%pip install pandas` in a cell once, then rerun. The download and preprocessing scripts need no extra packages.


```python
from pathlib import Path
import subprocess
import sys

project = Path.cwd()
if not (project / "download_data.py").is_file():
    raise RuntimeError("Open this notebook with the project folder as the working directory.")

raw_file = project / "data/raw/games.csv"
processed = project / "data/processed"
```

```python
if not raw_file.exists():
    subprocess.run([sys.executable, "download_data.py", "--version", "59"], check=True)
else:
    print("Raw data is already downloaded.")
```

```python
needed = ["cleaned_games.csv", "train.csv", "test.csv", "metadata.json"]
if not all((processed / name).exists() for name in needed):
    subprocess.run([sys.executable, "preprocess.py", "--input", str(raw_file)], check=True)
else:
    print("Processed data is already available.")
```

## View the cleaned games

Each row is one game with at least 100 reviews. `success=yes` means at least 80% of its reviews are positive. These thresholds are provisional. `games` below contains the inspection data; `train` and `test` contain the categorical columns used for modeling.

The raw file has a malformed header; preprocessing repairs it when reading. Use these processed tables for inspection. If preprocessing settings change, rerun `preprocess.py` to refresh the outputs before reloading the tables.


```python
import pandas as pd

games = pd.read_csv(processed / "cleaned_games.csv")
train = pd.read_csv(processed / "train.csv")
test = pd.read_csv(processed / "test.csv")

print(f"Eligible games: {len(games):,} | Train: {len(train):,} | Test: {len(test):,}")
games[["name", "price", "genres", "total_reviews", "positive_fraction", "success"]].head(10)
```
